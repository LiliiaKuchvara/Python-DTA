# Частина 1. Вступ до NumPy: створення масивів

## 1.1 Що таке масив

**Масив** (array) — це набір чисел одного типу. Найпростіше уявляти його як:
- **стовпчик в Excel** — одновимірний масив (наприклад, виручка по днях);
- **цілу таблицю** — двовимірний масив (наприклад, продажі магазинів по тижнях).

Масив створюють зі звичайного списку Python функцією `np.array()`.

````python
# Виручка інтернет-магазину за тиждень (грн): Пн, Вт, Ср, Чт, Пт, Сб, Нд
sales = np.array([12500, 13800, 11200, 15400, 18900, 22300, 19700])

print([12500, 13800, 11200, 15400, 18900, 22300, 19700])  # list
print(sales)  # array
print(type(sales))
````

## 1.2 Навіщо масив, якщо є звичайний список?

Головна суперсила масиву: **дія застосовується до кожного числа одразу**, без циклів. Порівняйте:

````python
prices_list = [100, 250, 80]            # звичайний список
prices_arr = np.array([100, 250, 80])   # масив NumPy

print("Список * 2:", prices_list * 2)   # список просто повторився двічі!
print("Масив  * 2:", prices_arr * 2)    # кожна ціна помножилась на 2
````

Тепер практичний приклад. Ціни на товари зросли на 10%, а ще треба показати їх у доларах:

````python
prices_uah = np.array([1200, 450, 3990, 860])   # ціни 4 товарів, грн

new_prices = prices_uah * 1.10                  # +10%
print("Нові ціни, грн:", new_prices)

usd_rate = 41.5                                 # курс долара (умовний)
print("Ціни в $:", np.round(prices_uah / usd_rate, 2))
````

Дії між **двома масивами однакової довжини** теж виконуються попарно: перше з першим, друге з другим і т.д.

````python
quantity = np.array([3, 10, 1, 5])      # скільки штук кожного товару продали

revenue = prices_uah * quantity         # виручка по кожному товару
print("Виручка по товарах:", revenue)
print("Загальна виручка:  ", revenue.sum())
````

## 1.5 Створення масивів «за шаблоном»

Часто масив не вводять вручну, а створюють готовою функцією:

| Функція | Що робить | Приклад для аналітика |
|---|---|---|
| `np.zeros(n)` | масив з нулів | порожня заготовка під план продажів |
| `np.ones(n)` | масив з одиниць | однакова «вага» для кожного клієнта |
| `np.full(n, x)` | масив, заповнений числом `x` | план 100 000 грн на кожен місяць |
| `np.arange(start, stop, step)` | послідовність з кроком | номери днів чи місяців |
| `np.linspace(start, stop, n)` | `n` рівномірних точок від `start` до `stop` | шкала знижок від 0% до 30% |

⚠️ У `np.arange` число `stop` **не входить** у результат: `np.arange(1, 13)` дає числа від 1 до 12.

````python
plan = np.zeros(12)                 # план на 12 місяців (поки що порожній)
print("zeros:", plan)

weights = np.ones(5)
print("ones: ", weights)

target = np.full(12, 100_000)       # 100_000 — це просто 100000, підкреслення для зручності читання
print("full: ", target)
````

````python
days = np.arange(1, 31)             # дні місяця: від 1 до 30 (31 не входить!)
print("Дні:     ", days)

months = np.arange(1, 13)           # місяці: від 1 до 12
print("Місяці:  ", months)

even_days = np.arange(2, 31, 2)     # кожен другий день (крок 2)
print("Парні дні:", even_days)

discounts = np.linspace(0, 30, 7)   # 7 рівних кроків від 0% до 30% (тут 30 входить!)
print("Знижки, %:", discounts)
````

## 1.6 Типи даних (dtype)

Усі числа в масиві мають **один тип**. Найчастіші:
- `int64` — цілі числа (кількість замовлень, штук, клієнтів);
- `float64` — дробові числа (ціни, середні значення, частки).

Тип можна змінити методом `.astype()`.

````python
orders = np.array([15, 22, 18])
avg_check = np.array([450.5, 380.0, 512.75])
print("orders:   ", orders.dtype)
print("avg_check:", avg_check.dtype)

# Якщо змішати цілі й дробові числа — NumPy зробить усі дробовими
mixed = np.array([1, 2, 3.5])
print("mixed:", mixed, mixed.dtype)

# Округлити суми до цілих гривень: спочатку .round(), потім .astype(int)
# (.astype(int) без округлення просто відкидає дробову частину: 512.75 → 512)
print(avg_check.round().astype(int))
````

## 1.7 Як дістати потрібні дані: індекси та зрізи

- Нумерація починається з **0**: перший елемент — `[0]`, другий — `[1]`.
- `[-1]` — останній елемент, `[-2]` — передостанній.
- Зріз `[старт:стоп]` бере елементи від `старт` до `стоп`, **не включаючи** `стоп`.

````python
print("Понеділок:", sales[0])
print("Неділя:   ", sales[-1])
print("Пн–Ср:    ", sales[0:3])    # елементи 0, 1, 2
print("Вихідні:  ", sales[-2:])    # два останні
````

Для таблиці пишемо `[рядок, стовпець]`. Двокрапка `:` означає «усі».

````python
print(store_sales, "\n")

print("Київ, 1-й тиждень:        ", store_sales[0, 0])
print("Львів, усі тижні:         ", store_sales[1, :])
print("Усі магазини, 4-й тиждень:", store_sales[:, 3])
print("2-3-й магазини, 2-3-й тижні:\n", store_sales[1:3, 1:3])
````

## 1.8 Фільтрація — як фільтр в Excel

Порівняння (`>`, `<`, `==`) дає масив з `True`/`False` — його називають **маскою**. Якщо підставити маску у квадратні дужки, залишаться лише ті значення, де `True`.

````python
mask = sales > 15000

print("Sales: ", sales)
print("Маска:", mask)
print("Дні з виручкою понад 15 000:", sales[mask])
print("Скільки таких днів:", mask.sum())   # True рахується як 1, False як 0
````

Кілька умов поєднують через `&` («і») та `|` («або»). Кожну умову беремо **в дужки**.

````python
# Дні з виручкою від 12 000 до 16 000
print(sales[(sales >= 12000) & (sales <= 16000)])

# Дні з дуже низькою АБО дуже високою виручкою
print(sales[(sales < 12000) | (sales > 20000)])
````

## 1.9 Підсумки: sum, mean, min, max

Основні «підсумкові» функції аналітика:

````python
day_names = np.array(["Пн", "Вт", "Ср", "Чт", "Пт", "Сб", "Нд"])

print("Сума за тиждень:", sales.sum())
print("Середнє за день:", sales.mean().round(1))
print("Мінімум:", sales.min(), "| Максимум:", sales.max())

best = sales.argmax()                     # argmax — НОМЕР (індекс) найбільшого значення
print("Найкращий день:", day_names[best])
````

Для таблиці важливий параметр **`axis`** — «в якому напрямку рахувати»:

- `axis=1` — рахуємо **вздовж рядка** → отримуємо підсумок для кожного рядка (кожного магазину);
- `axis=0` — рахуємо **вниз по стовпцю** → отримуємо підсумок для кожного стовпця (кожного тижня).

```
                тиждень 1   2    3    4   │ axis=1 (сума рядка)
Київ               120     135  128  150  │  →  533
Львів               80      95   90  110  │  →  375
Одеса               60      58   72   75  │  →  265
──────────────────────────────────────────┘
axis=0 (сума       260     288  290  335
стовпця) ↓
```

````python
print("По магазинах (axis=1):", store_sales.sum(axis=1))
print("По тижнях    (axis=0):", store_sales.sum(axis=0))
print("Разом:", store_sales.sum())
print("Середній тиждень кожного магазину:", store_sales.mean(axis=1).round(1))
````

Ще дві корисні функції:
- `cumsum()` — **накопичувальний підсумок** (скільки заробили від початку тижня);
- `np.diff()` — **зміна** порівняно з попереднім днем.

````python
print("Виручка:                  ", sales)
print("Накопичена виручка:       ", sales.cumsum())
print("Зміна до попереднього дня:", np.diff(sales))
````

# Частина 2. Лінійна алгебра

Не лякайтеся назви! Для аналітика лінійна алгебра — це здебільшого одна дія:
### **«помножити кількість на ціну і все скласти»**

Саме так рахують суму чека, виручку, собівартість. NumPy робить це для тисяч клієнтів і товарів **одним рядком**.

Два терміни, які знадобляться:
- **Вектор** — одновимірний масив (наприклад, ціни товарів).
- **Матриця** — двовимірний масив, таблиця (наприклад, замовлення: клієнти × товари).

## 2.1 Скалярний добуток: сума одного чека

Клієнт купив 2 кави, 1 круасан і 3 печива. Скільки він заплатив?

Рахуємо як у школі: `2·60 + 1·70 + 3·25 = 265`. Ця дія називається **скалярний добуток** (dot product). У NumPy є три способи:

````python
quantity = np.array([2, 1, 3])      # кава, круасан, печиво (шт)
prices = np.array([60, 70, 25])     # ціна за штуку (грн)

# Спосіб 1: помножити попарно і скласти
print("Спосіб 1:", (quantity * prices).sum())

# Спосіб 2: функція np.dot
print("Спосіб 2:", np.dot(quantity, prices))

# Спосіб 3: оператор @ — найкоротший, ним і будемо користуватись
print("Спосіб 3:", quantity @ prices)
````

## 2.2 Матриця × вектор: чеки всіх клієнтів одразу

Тепер у нас не один клієнт, а чотири. Записуємо їхні замовлення в таблицю (рядки — клієнти, стовпці — товари) і множимо на ціни через `@`.

Що відбувається: **кожен рядок** таблиці множиться на вектор цін так само, як у 2.1.

````python
clients = np.array(["Олена", "Андрій", "Марія", "Ігор"])

#                кава круасан печиво
orders = np.array([[2,   1,     3],    # Олена
                   [1,   0,     0],    # Андрій
                   [0,   2,     4],    # Марія
                   [3,   3,     1]])   # Ігор

checks = orders @ prices
print("Суми чеків:", checks)

for name, check in zip(clients, checks):
    print(f"  {name}: {check} грн")
````

## 2.3 Матриця × матриця: виручка і собівартість по магазинах

Мережа з 3 кав'ярень. Знаємо, скільки кожного товару продав кожен магазин, а також ціну і собівартість кожного товару. Множимо одну таблицю на іншу — і отримуємо **виручку та собівартість для кожного магазину**.

````python
# Продано штук: рядки — магазини, стовпці — товари
#                кава круасан печиво
sold = np.array([[300, 120,   200],    # Київ
                 [250, 100,   150],    # Львів
                 [180,  60,   120]])   # Одеса

# Для кожного товару: [ціна продажу, собівартість]
unit_economics = np.array([[60, 22],    # кава
                           [70, 35],    # круасан
                           [25, 10]])   # печиво

result = sold @ unit_economics
print("Стовпці: виручка, собівартість")
print(result)
````

````python
revenue = result[:, 0]     # перший стовпець — виручка
cost = result[:, 1]        # другий стовпець — собівартість
profit = revenue - cost

print("Прибуток по магазинах:", profit)
print("Маржа, %:             ", np.round(profit / revenue * 100, 1))
````

### Головне правило множення таблиць

Щоб помножити `A @ B`, **кількість стовпців A має дорівнювати кількості рядків B**. Ці «внутрішні» розміри зникають, а результат має «зовнішні» розміри:

```
   sold          @   unit_economics   =   result
(3 магазини × 3 товари) @ (3 товари × 2 показники) = (3 магазини × 2 показники)
              └──── мають збігатися ────┘
```

У нашому прикладі внутрішній розмір — це **товари**: вони є і в першій, і в другій таблиці.

````python
print(sold.shape, "@", unit_economics.shape, "=", result.shape)
````

## 2.4 Транспонування: «перевернути» таблицю

`.T` міняє рядки і стовпці місцями. Це рятує, коли дані «лежать не так», як потрібно для множення.

````python
print("Було (товари × показники):")
print(unit_economics)
print("\nСтало (показники × товари):")
print(unit_economics.T)
print("\nФорма:", unit_economics.shape, "→", unit_economics.T.shape)
````

Приклад: колега надіслав ціни та собівартість «горизонтально» — у вигляді 2 рядків. Пряме множення не спрацює, бо розміри не збігаються. Транспонуємо — і все працює.

````python
#                     кава круасан печиво
econ_wide = np.array([[60,  70,    25],    # ціна
                      [22,  35,    10]])   # собівартість

try:
    sold @ econ_wide
except ValueError:
    print("❌ Помилка! Розміри не підходять:", sold.shape, "@", econ_wide.shape)

print("✅ Після транспонування:", sold.shape, "@", econ_wide.T.shape)
print(sold @ econ_wide.T)
````

## 2.5 Система рівнянь: знаходимо невідомі ціни

**Ситуація:** у вас є два чеки, але немає прайсу.
- Чек 1: 2 кави + 1 десерт = **190 грн**
- Чек 2: 1 кава + 3 десерти = **270 грн**

Скільки коштує кава, а скільки десерт? Це **система рівнянь**, і `np.linalg.solve` розв'яже її за нас:
- у таблицю `A` записуємо кількості з чеків;
- у вектор `b` — суми чеків.

````python
#             кава десерт
A = np.array([[2,  1],      # чек 1
              [1,  3]])     # чек 2
b = np.array([190, 270])    # суми чеків

found_prices = np.linalg.solve(A, b)
print("Кава:  ", found_prices[0].round(2), "грн")
print("Десерт:", found_prices[1].round(2), "грн")

# Перевірка: множимо кількості на знайдені ціни — маємо отримати суми чеків
print("Перевірка:", A @ found_prices)
````

💡 Щоб це працювало, **рівнянь має бути стільки ж, скільки невідомих**, і вони не повинні «повторювати» одне одного. Наприклад, якщо другий чек — це просто подвоєний перший (4 кави + 2 десерти = 380), нової інформації він не дає, і NumPy видасть помилку `LinAlgError`.

# Частина 3. Випадкові числа

Навіщо аналітику випадкові числа?
- створити **тестові дані**, поки справжніх ще немає;
- зробити **випадкову вибірку** клієнтів (наприклад, для опитування);
- розділити користувачів на групи **A/B-тесту**;
- **змоделювати** майбутнє: «яка виручка можлива в гіршому і кращому випадку?»

## 3.1 Генератор і seed

Випадкові числа в NumPy видає **генератор**: `rng = np.random.default_rng(seed)`.

**Seed** («зерно») — будь-яке число, яке робить «випадковість» **відтворюваною**: з тим самим seed ви щоразу отримаєте ті самі числа. Це важливо, щоб колега міг повторити ваш аналіз, а у вас вийшли ті самі результати, що в цьому ноутбуці.

````python
rng_a = np.random.default_rng(7)
rng_b = np.random.default_rng(7)

print(rng_a.integers(0, 100, size=5))
print(rng_b.integers(0, 100, size=5))   # той самий seed → ті самі числа
````

💡 В інших курсах ви можете побачити `np.random.seed(...)` та `np.random.randint(...)`. Це **старий спосіб** — він працює, але зараз рекомендують `default_rng`.

У кожній клітинці нижче ми створюємо генератор заново з `seed=42` — тож скільки б разів ви не перезапускали клітинку, результат буде той самий.

## 3.2 Цілі випадкові числа: `integers`

`rng.integers(low, high, size)` — цілі числа від `low` до `high`, **не включаючи** `high` (як у `np.arange`).

````python
rng = np.random.default_rng(42)

# Кількість замовлень на день за 2 тижні: від 20 до 50
daily_orders = rng.integers(20, 51, size=14)
print("Замовлення по днях:", daily_orders)
print("Середнє:", daily_orders.mean().round(1))

# size може бути таблицею: 3 магазини × 7 днів
week_orders = rng.integers(20, 51, size=(3, 7))
print(week_orders)
````

## 3.3 Дробові числа: `random` і `uniform`

- `rng.random(size)` — числа від 0 до 1 (зручно для часток і ймовірностей);
- `rng.uniform(low, high, size)` — будь-які числа від `low` до `high`, усі однаково ймовірні.

````python
rng = np.random.default_rng(42)

print("Від 0 до 1:", rng.random(5).round(3))

test_prices = rng.uniform(100, 500, size=5).round(2)   # тестові ціни від 100 до 500 грн
print("Ціни:", test_prices)
````

## 3.4 Нормальний розподіл: «більшість значень біля середнього»

Багато бізнес-показників (середній чек, час доставки, зріст покупців) групуються навколо середнього: значень біля середнього багато, а дуже малих чи дуже великих — мало. Це **нормальний розподіл** — графік у формі «дзвона».

`rng.normal(loc=середнє, scale=стандартне_відхилення, size=кількість)`

**Стандартне відхилення** (std) показує, наскільки сильно значення «розкидані» навколо середнього. Корисне правило:
- приблизно **68%** значень лежать у межах **середнє ± 1 std**;
- приблизно **95%** — у межах **середнє ± 2 std**.

````python
rng = np.random.default_rng(42)

# 1000 чеків: середній чек 450 грн, стандартне відхилення 80 грн
checks = rng.normal(loc=450, scale=80, size=1000).round(2)

print("Перші 5 чеків:", checks[:5])
print("Середній чек:", checks.mean().round(2))
print("Стандартне відхилення:", checks.std().round(2))

# Частка чеків від 290 до 610 грн (450 ± 2·80).
# Середнє з True/False = частка True
share = ((checks > 290) & (checks < 610)).mean()
print("Частка чеків у межах ±2 std:", (share * 100).round(1), "%")
````

````python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 4))
plt.hist(checks, bins=30, color="#2a78d6", edgecolor="white", linewidth=1.5)
plt.axvline(checks.mean(), color="#0b0b0b", linestyle="--", linewidth=1.5,
            label=f"Середнє: {checks.mean():.0f} грн")
plt.title("Розподіл 1000 чеків")
plt.xlabel("Сума чека, грн")
plt.ylabel("Кількість чеків")
plt.grid(axis="y", alpha=0.3)
plt.gca().spines[["top", "right"]].set_visible(False)
plt.legend(frameon=False)
plt.show()
````

## 3.5 Випадковий вибір: `choice`

`rng.choice(масив, size, replace=False)` випадково обирає елементи з масиву. `replace=False` означає «**без повторів**» — один клієнт не потрапить у вибірку двічі.

````python
rng = np.random.default_rng(42)

client_ids = np.arange(1001, 1021)   # ID 20 клієнтів: 1001 ... 1020
survey = rng.choice(client_ids, size=5, replace=False)
print("Клієнти для опитування:", survey)
````

Можна задати **ймовірності** через параметр `p` (їхня сума має дорівнювати 1). Змоделюємо 1000 візитів на сайт з різних каналів, а потім порахуємо їх функцією `np.unique(..., return_counts=True)`:

````python
rng = np.random.default_rng(42)

channels = np.array(["Google", "Instagram", "Email", "Direct"])
visits = rng.choice(channels, size=1000, p=[0.4, 0.3, 0.1, 0.2])

print("Перші 10 візитів:", visits[:10])

names, counts = np.unique(visits, return_counts=True)
print("\nВізитів з кожного каналу:")
for name, count in zip(names, counts):
    print(f"  {name:<10} {count}")
````

## 3.6 Групи для A/B-тесту: `permutation`

`rng.permutation(масив)` випадково **перемішує** елементи. Перемішали — і розрізали навпіл: отримали дві випадкові групи.

````python
rng = np.random.default_rng(42)

users = np.arange(1, 11)           # 10 користувачів з ID від 1 до 10
shuffled = rng.permutation(users)
print("Перемішані:", shuffled)

group_a = shuffled[:5]
group_b = shuffled[5:]
print("Група A:", np.sort(group_a))
print("Група B:", np.sort(group_b))
````

## 3.7 Скільки людей купить? Біноміальний розподіл

Сайт щодня відвідує 1000 людей, конверсія — 3%. В середньому це 30 покупок, але кожного дня трохи по-різному. `rng.binomial(n, p, size)` моделює саме таку ситуацію: `n` спроб, кожна вдала з ймовірністю `p`.

````python
rng = np.random.default_rng(42)

purchases = rng.binomial(n=1000, p=0.03, size=7)
print("Покупки за тиждень по днях:", purchases)
print("Середнє:", purchases.mean().round(1))
````

## 3.8 Міні-прогноз: виручка на місяць (метод Монте-Карло)

Назва звучить складно, але ідея проста: **«програємо» місяць 10 000 разів** з випадковими даними і дивимось, які результати трапляються найчастіше.

Припущення:
- щодня 1000 відвідувачів, конверсія 3%;
- середній чек дня коливається біля 450 грн (std 30 грн).

Тут знадобиться все, що ми вивчили: таблиці (10 000 місяців × 30 днів), множення масивів, `sum(axis=1)`.

**Перцентиль** — це «межа»: 5-й перцентиль означає, що лише в 5% симуляцій виручка була нижчою. Тобто це **песимістичний, але реалістичний** сценарій.

````python
rng = np.random.default_rng(42)

n_sims = 10_000   # скільки місяців «програємо»
n_days = 30

orders_sim = rng.binomial(n=1000, p=0.03, size=(n_sims, n_days))     # замовлення по днях
check_sim = rng.normal(loc=450, scale=30, size=(n_sims, n_days))     # середній чек по днях

month_revenue = (orders_sim * check_sim).sum(axis=1)                 # виручка кожного місяця

print(f"Очікувана виручка:            {month_revenue.mean() / 1000:.1f} тис. грн")
print(f"Песимістичний сценарій (5%):  {np.percentile(month_revenue, 5) / 1000:.1f} тис. грн")
print(f"Оптимістичний сценарій (95%): {np.percentile(month_revenue, 95) / 1000:.1f} тис. грн")
````

````python
plt.figure(figsize=(8, 4))
plt.hist(month_revenue / 1000, bins=40, color="#2a78d6", edgecolor="white", linewidth=1)
for q in (5, 95):
    plt.axvline(np.percentile(month_revenue, q) / 1000, color="#0b0b0b",
                linestyle="--", linewidth=1.5)
plt.title("10 000 симуляцій місячної виручки (пунктир — 5% і 95%)")
plt.xlabel("Виручка за місяць, тис. грн")
plt.ylabel("Кількість симуляцій")
plt.grid(axis="y", alpha=0.3)
plt.gca().spines[["top", "right"]].set_visible(False)
plt.show()
````

# 🎁 Бонус: усе разом — скільки продажів дає реклама?

Згенеруємо тестові дані за рік (52 тижні): бюджет на рекламу і продажі. Продажі залежать від реклами, але з випадковим «шумом» (погода, сезон, конкуренти — усе, що ми не контролюємо).

Потім знайдемо **лінію тренду** функцією `np.polyfit` (всередині вона використовує ту саму лінійну алгебру) і подивимось, чи зможемо «вгадати» закладену залежність.

````python
rng = np.random.default_rng(42)

ads = rng.uniform(5, 50, size=52).round(1)        # бюджет на рекламу, тис. грн
noise = rng.normal(0, 12, size=52)                # випадковий «шум»
weekly_sales = 50 + 3.2 * ads + noise             # закладена залежність: 50 + 3.2 тис. на кожну 1 тис. реклами

slope, intercept = np.polyfit(ads, weekly_sales, deg=1)   # deg=1 — пряма лінія
print(f"Кожна 1 тис. грн реклами ≈ +{slope:.2f} тис. грн продажів")
print(f"Продажі без реклами ≈ {intercept:.1f} тис. грн")

line_x = np.linspace(ads.min(), ads.max(), 50)
plt.figure(figsize=(8, 4))
plt.scatter(ads, weekly_sales, s=40, color="#2a78d6", edgecolor="white", linewidth=1.5, label="Тижні")
plt.plot(line_x, intercept + slope * line_x, color="#eb6834", linewidth=2, label="Лінія тренду")
plt.title("Реклама і продажі за 52 тижні")
plt.xlabel("Бюджет на рекламу, тис. грн")
plt.ylabel("Продажі, тис. грн")
plt.grid(alpha=0.3)
plt.gca().spines[["top", "right"]].set_visible(False)
plt.legend(frameon=False)
plt.show()
````

Модель знайшла числа, близькі до закладених (50 і 3.2), хоча «бачила» лише зашумлені дані. Точно вони не збігаються саме через шум — і чим більше даних, тим точніша оцінка. Так аналітики оцінюють ефект реклами на справжніх даних.

# 📋 Шпаргалка

**Створення масивів**

| Код | Що робить |
|---|---|
| `np.array([1, 2, 3])` | масив зі списку |
| `np.zeros(n)`, `np.ones(n)`, `np.full(n, x)` | масив з нулів / одиниць / числа x |
| `np.arange(1, 13)` | числа від 1 до 12 |
| `np.linspace(0, 30, 7)` | 7 рівних кроків від 0 до 30 |
| `a.shape`, `a.dtype`, `a.size` | форма, тип, кількість елементів |
| `a.reshape(4, 3)` | змінити форму |
| `a.astype(int)` | змінити тип |

**Вибір і підсумки**

| Код | Що робить |
|---|---|
| `a[0]`, `a[-1]`, `a[1:4]` | перший, останній, зріз |
| `t[рядок, стовпець]`, `t[:, 0]` | елемент таблиці, цілий стовпець |
| `a[a > 100]` | фільтр |
| `a.sum()`, `a.mean()`, `a.min()`, `a.max()` | підсумки |
| `t.sum(axis=0)` / `t.sum(axis=1)` | по стовпцях / по рядках |
| `a.argmax()`, `a.cumsum()`, `np.diff(a)` | індекс максимуму, накопичений підсумок, зміна |

**Лінійна алгебра**

| Код | Що робить |
|---|---|
| `a @ b` | кількість × ціна і скласти (чек, виручка) |
| `t.T` | перевернути таблицю |
| `np.linalg.solve(A, b)` | розв'язати систему рівнянь |

**Випадкові числа**

| Код | Що робить |
|---|---|
| `rng = np.random.default_rng(42)` | генератор з seed |
| `rng.integers(low, high, size)` | цілі числа (high не включається) |
| `rng.random(size)`, `rng.uniform(low, high, size)` | дробові числа |
| `rng.normal(середнє, std, size)` | нормальний розподіл |
| `rng.choice(a, size, replace=False, p=...)` | випадковий вибір |
| `rng.permutation(a)` | перемішати |
| `rng.binomial(n, p, size)` | кількість «успіхів» (покупок) |
| `np.percentile(a, 5)` | 5-й перцентиль |

### Що далі?
Наступний крок — **pandas**. Він побудований на NumPy, тому все, що ви тут вивчили (фільтри, `sum(axis=...)`, `mean`, випадкові дані), там працює майже так само — тільки з назвами стовпців і зручним виглядом таблиць.
