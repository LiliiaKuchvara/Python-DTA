## Завдання 1. Перевірка та очищення даних

Переглянь усі поїздки та розподіли їх:
- скасовані поїздки;
- **некоректні** завершені поїздки: відстань дорівнює 0 або менше — причина `"invalid distance"`; оцінка не є цілим числом від 1 до 5 — причина `"invalid rating"`. Якщо в поїздці кілька проблем, вказуй лише першу з них у цьому порядку;
- **коректні** завершені поїздки — збережи їх в окремий список `valid_rides`, з ним ти працюватимеш далі.

Виведи загальну кількість поїздок, кількість скасованих, кількість коректних, а також номер і причину для кожної некоректної поїздки

```python

# Створюємо списки для розподілу поїздок
cancelled_rides = []
invalid_rides = []
valid_rides = []


for ride in rides:
    if ride["status"] == "cancelled":
        cancelled_rides.append(ride)

    elif ride["status"] == "completed":
        if ride["distance"] <= 0:
            invalid_rides.append({
                "id": ride["id"],
                "reason": "invalid distance"
            })
        elif not isinstance(ride["rating"], int) or not 1 <= ride["rating"] <= 5:
            invalid_rides.append({
                "id": ride["id"],
                "reason": "invalid rating"
            })
        else:
            valid_rides.append(ride)


print("Загальна кількість поїздок:", len(rides))
print("Кількість скасованих:", len(cancelled_rides))
print("Кількість коректних:", len(valid_rides))

print("\nНекоректні поїздки:")
for ride in invalid_rides:
    print(f'Поїздка {ride["id"]}: {ride["reason"]}')
```

## Завдання 2. Розрахунок вартості поїздок

Для кожної поїздки з `valid_rides` розрахуй вартість за тарифом:
- посадка — 50 грн плюс 12 грн за кожен кілометр;
- якщо поїздка тривала понад 30 хвилин — додатково 3 грн за кожну хвилину понад 30;
- нічний тариф (поїздка почалася з 22:00 до 05:59 включно) — уся вартість збільшується на 25%;
- мінімальна вартість поїздки — 80 грн: якщо розрахована сума менша, пасажир платить 80 грн.

Округли вартість до 2 знаків і додай її до словника кожної поїздки з ключем `"price"`. Виведи для кожної поїздки номер, водія, відстань і вартість

```python

for ride in valid_rides:
    # Базова вартість: посадка + оплата за відстань
    price = 50 + ride["distance"] * 12

    # Доплата за тривалість понад 30 хвилин
    if ride["duration"] > 30:
        price += (ride["duration"] - 30) * 3

    # Нічний тариф: з 22:00 до 05:59
    if ride["hour"] >= 22 or ride["hour"] < 6:
        price *= 1.25

    # Мінімальна вартість — 80 грн
    if price < 80:
        price = 80

    # Округлюємо до 2 знаків після коми
    ride["price"] = round(price, 2)

# Виводимо результати
for ride in valid_rides:
    print(
        f'Поїздка {ride["id"]}: '
        f'водій — {ride["driver"]}, '
        f'відстань — {ride["distance"]} км, '
        f'вартість — {ride["price"]:.2f} грн'
    )
```

## Завдання 3. Статистика по водіях

На основі `valid_rides` створи словник `driver_stats`, у якому для кожного водія зберігаються:
- кількість поїздок;
- виручка (сума вартостей його поїздок, округлена до 2 знаків);
- середній рейтинг, округлений до 1 знака.

Виведи статистику кожного водія в одному рядку

```python

driver_stats = {}

# Збираємо статистику для кожного водія
for ride in valid_rides:
    driver = ride["driver"]

    if driver not in driver_stats:
        driver_stats[driver] = {
            "rides_count": 0,
            "revenue": 0,
            "ratings_sum": 0,
            "ratings_count": 0
        }

    driver_stats[driver]["rides_count"] += 1
    driver_stats[driver]["revenue"] += ride["price"]
    driver_stats[driver]["ratings_sum"] += ride["rating"]
    driver_stats[driver]["ratings_count"] += 1

# Розраховуємо виручку та середній рейтинг
for driver, stats in driver_stats.items():
    stats["revenue"] = round(stats["revenue"], 2)
    stats["average_rating"] = round(
        stats["ratings_sum"] / stats["ratings_count"], 1
    )

# Виводимо статистику кожного водія в одному рядку
for driver, stats in driver_stats.items():
    print(
        f'{driver}: '
        f'поїздок — {stats["rides_count"]}, '
        f'виручка — {stats["revenue"]:.2f} грн, '
        f'середній рейтинг — {stats["average_rating"]:.1f}'
    )
```

## Завдання 4. Статуси та бонуси водіїв

На основі `driver_stats` визнач для кожного водія статус і бонус (середній рейтинг береться округленим до 1 знака):
- рейтинг від 4.8 **і** щонайменше 3 поїздки — статус `"Gold"`, бонус 10% від виручки;
- рейтинг від 4.5 — статус `"Silver"`, бонус 5% від виручки;
- рейтинг нижче 4.0 — статус `"Warning"`, бонусу немає;
- в інших випадках — статус `"Standard"`, бонусу немає.

Округли бонус до 2 знаків і додай статус та бонус до статистики водія. Виведи імʼя, статус і бонус кожного водія, а наприкінці — загальну суму бонусів

```python

total_bonus = 0

for driver, stats in driver_stats.items():
    rating = stats["average_rating"]
    rides_count = stats["rides_count"]
    revenue = stats["revenue"]

    # Визначаємо статус і бонус
    if rating >= 4.8 and rides_count >= 3:
        status = "Gold"
        bonus = revenue * 0.10
    elif rating >= 4.5:
        status = "Silver"
        bonus = revenue * 0.05
    elif rating < 4.0:
        status = "Warning"
        bonus = 0
    else:
        status = "Standard"
        bonus = 0

    # Округлюємо бонус і зберігаємо результати
    bonus = round(bonus, 2)

    stats["status"] = status
    stats["bonus"] = bonus

    total_bonus += bonus

    print(
        f'{driver}: статус — {status}, '
        f'бонус — {bonus:.2f} грн'
    )

print(f"\nЗагальна сума бонусів: {total_bonus:.2f} грн")
```

## Завдання 5. Підсумковий звіт за день

Виведи звіт, у якому кожен показник займає окремий рядок:
- загальна виручка за день;
- кількість коректних поїздок і середня вартість поїздки (округли до 2 знаків);
- частка скасованих поїздок від усіх поїздок у журналі, % (округли до 1 знака);
- частка поїздок, оплачених карткою, серед коректних поїздок, % (округли до 1 знака);
- найзавантаженіша година — година, на яку припало найбільше коректних поїздок, і кількість поїздок у цю годину;
- найдовша поїздка: номер, водій і відстань;
- водій з найбільшою виручкою;
- загальна сума бонусів водіям.

Не використовуй функції `max()` і `min()`

```python

# 1. Загальна виручка за день
total_revenue = 0

for ride in valid_rides:
    total_revenue += ride["price"]

total_revenue = round(total_revenue, 2)

# 2. Кількість коректних поїздок і середня вартість
rides_count = len(valid_rides)

if rides_count > 0:
    average_price = round(total_revenue / rides_count, 2)
else:
    average_price = 0

# 3. Частка скасованих поїздок
cancelled_count = len(cancelled_rides)

if len(rides) > 0:
    cancelled_percent = round(cancelled_count / len(rides) * 100, 1)
else:
    cancelled_percent = 0

# 4. Частка оплат карткою серед коректних поїздок
card_count = 0

for ride in valid_rides:
    if ride["payment"] == "card":
        card_count += 1

if rides_count > 0:
    card_percent = round(card_count / rides_count * 100, 1)
else:
    card_percent = 0

# 5. Найзавантаженіша година
hour_counts = {}

for ride in valid_rides:
    hour = ride["hour"]

    if hour not in hour_counts:
        hour_counts[hour] = 0

    hour_counts[hour] += 1

busiest_hour = None
busiest_count = 0

for hour, count in hour_counts.items():
    if count > busiest_count:
        busiest_hour = hour
        busiest_count = count

# 6. Найдовша поїздка
longest_ride = None

for ride in valid_rides:
    if longest_ride is None or ride["distance"] > longest_ride["distance"]:
        longest_ride = ride

# 7. Водій з найбільшою виручкою
top_driver = None
top_revenue = 0

for driver, stats in driver_stats.items():
    if top_driver is None or stats["revenue"] > top_revenue:
        top_driver = driver
        top_revenue = stats["revenue"]

# 8. Загальна сума бонусів
total_bonus = 0

for stats in driver_stats.values():
    total_bonus += stats["bonus"]

total_bonus = round(total_bonus, 2)

# Підсумковий звіт
print("ПІДСУМКОВИЙ ЗВІТ ЗА ДЕНЬ")
print("-" * 40)
print(f"Загальна виручка: {total_revenue:.2f} грн")
print(f"Кількість коректних поїздок: {rides_count}")
print(f"Середня вартість поїздки: {average_price:.2f} грн")
print(f"Частка скасованих поїздок: {cancelled_percent:.1f}%")
print(f"Оплата карткою: {card_percent:.1f}%")
print(
    f"Найзавантаженіша година: {busiest_hour}:00 "
    f"({busiest_count} поїздки)"
)
print(
    f'Найдовша поїздка: №{longest_ride["id"]}, '
    f'{longest_ride["driver"]}, '
    f'{longest_ride["distance"]:.1f} км'
)
print(f"Водій з найбільшою виручкою: {top_driver} ({top_revenue:.2f} грн)")
print(f"Загальна сума бонусів: {total_bonus:.2f} грн")
```
