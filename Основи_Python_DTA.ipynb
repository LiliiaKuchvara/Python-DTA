## Змінні

**Змінна** — це «коробка з наліпкою»: ми кладемо в неї значення і звертаємося до нього за назвою.

```
назва = значення
```

Знак `=` означає «**поклади** значення в змінну», а не «дорівнює», як у математиці.

````python
coffee_price = 65
cups = 3
print(coffee_price * cups)       # Python бере значення зі змінних

cafe_name = "Зерно"
print("Ласкаво просимо до кав'ярні", cafe_name)
````

Змінній можна дати **нове значення** — старе зникне. Нове значення можна обчислити зі старого:

````python
cups = 3
print("Було:", cups)

cups = cups + 2       # беремо старе значення (3), додаємо 2 і кладемо назад
print("Стало:", cups)
````

**Правила назв змінних**

| Назва | Можна? | Чому |
|---|---|---|
| `coffee_price` | ✅ | малі латинські літери, слова через `_` |
| `price2` | ✅ | цифра — не на початку |
| `2price` | ❌ | не можна починати з цифри |
| `coffee-price` | ❌ | дефіс Python сприйме як мінус |
| `coffee price` | ❌ | пробілів у назві бути не може |
| `print` | ⚠️ | можна, але так ти «зламаєш» функцію `print` |

Великі й малі літери — це різні назви: `Price` і `price` — дві різні змінні.

# Тема 2. Типи даних і перетворення
### Datentypen und Typumwandlung

| Deutsch | Українська |
|---|---|
| die Ganzzahl | ціле число |
| die Kommazahl | дробове число |

## Чотири базові типи

| Тип | Що зберігає | Приклад |
|---|---|---|
| `int` | ціле число | `3`, `-10`, `2026` |
| `float` | дробове число (через **крапку**!) | `64.5`, `0.2` |
| `str` | рядок — текст у лапках | `"Лате"`, `'15'` |
| `bool` | логічне значення | `True`, `False` |

Дізнатися тип допомагає `type()`.

````python
cups = 3              # int
price = 64.5          # float
drink = "Лате"        # str
is_open = True        # bool

print(type(cups))
print(type(price))
print(type(drink))
print(type(is_open))
````

## Чому тип важливий

Те саме «5» поводиться по-різному залежно від типу: числа **додаються**, а рядки **склеюються**.

````python
print(5 + 5)        # 10 — числа
print("5" + "5")    # 55 — рядки склеїлися
print("5" * 3)      # 555 — рядок повторився тричі
````

## Перетворення типів

| Функція | Що робить | Приклад |
|---|---|---|
| `int()` | у ціле число | `int("42")` → `42`, `int(3.99)` → `3` |
| `float()` | у дробове | `float("64.5")` → `64.5` |
| `str()` | у рядок | `str(2026)` → `"2026"` |
| `bool()` | у логічне | `bool(0)` → `False` |

Зверни увагу: `int(3.99)` дає `3` — дробова частина **відкидається**, а не округлюється.

````python
text = "42"
number = int(text)
print(number + 8)

print(float("64.5") * 2)
print(str(2026) + " рік")
print(int(3.99))
````

**Типові помилки:**
- `int("сорок")` → `ValueError`: Python не вміє перетворити такий текст на число.
- `int("64.5")` → `ValueError`: рядок з крапкою спершу перетворюють на `float`.
- `"Ціна: " + 65` → `TypeError`: не можна склеїти текст і число. Правильно: `"Ціна: " + str(65)`.

## Індекси та зрізи

Кожен символ має номер — **індекс**. Відлік починається з **нуля**. Від'ємні індекси рахують з кінця: `-1` — останній символ.

```
 К   а   в   а
 0   1   2   3
-4  -3  -2  -1
```

**Зріз** `text[start:stop]` бере символи від `start` **до** `stop` (сам `stop` не входить).

````python
word = "Капучино"
print(word[0])        # К
print(word[-1])       # о
print(word[0:4])      # Капу — індекси 0, 1, 2, 3
print(word[:4])       # те саме: початок можна не писати
print(word[4:])       # чино — до кінця
````

## Корисні методи

Метод — це дія, яку вміє робити рядок. Пишемо через крапку: `text.upper()`. Методи **не змінюють** рядок, а повертають новий.

| Метод | Що робить | Приклад → результат |
|---|---|---|
| `.upper()` / `.lower()` | ВЕЛИКІ / малі літери | `"Лате".upper()` → `"ЛАТЕ"` |
| `.capitalize()` | перша літера велика | `"лате"` → `"Лате"` |
| `.strip()` | прибирає пробіли по краях | `"  чай "` → `"чай"` |
| `.replace(a, b)` | замінює `a` на `b` | `"кава".replace("к", "К")` → `"Кава"` |
| `.count(a)` | скільки разів трапляється `a` | `"лате, лате".count("лате")` → `2` |
| `.startswith(a)` | чи починається з `a` | `"UA-001".startswith("UA")` → `True` |

````python
raw = "   кАВА з молоком  "
print(raw.strip())
print(raw.strip().capitalize())
print(raw.upper())
print("лате, чай, лате".count("лате"))
````

| `!=` | не дорівнює | `5 != 3` → `True` |
| `>` / `<` | більше / менше | `7 > 10` → `False` |
| `>=` / `<=` | більше або дорівнює / менше або дорівнює | `18 >= 18` → `True` |

⚠️ Не плутай: `=` — **покласти** значення в змінну, `==` — **порівняти**.

````python
age = 17
print(age >= 18)
print("лате" == "Лате")       # великі й малі літери різні
print(10 != 3)
````

## `and`, `or`, `not`

| Оператор | Коли `True` |
|---|---|
| `and` | обидві умови істинні |
| `or` | хоча б одна умова істинна |
| `not` | перевертає значення: `not True` → `False` |

````python
is_member = True
order_sum = 250

print(is_member and order_sum > 300)   # False: друга умова хибна
print(is_member or order_sum > 300)    # True: перша умова істинна
print(not is_member)                   # False
print(8 <= 14 < 20)                    # можна писати «подвійне» порівняння
````

# Тема 8. Цикл `while`, `break` і `continue`
### Die while-Schleife

| Deutsch | Українська |
|---|---|
| solange | доки (`while`) |
| die Endlosschleife | нескінченний цикл |
| abbrechen | перервати (`break`) |
| überspringen | пропустити (`continue`) |

## `while` — повторювати, доки умова істинна

```
while умова:
    код
```

Python перевіряє умову перед кожним повтором. Щойно вона стає `False` — цикл закінчується.

**Коли що обирати:**
- `for` — коли відомо, **що** перебирати (список, `range`);
- `while` — коли невідомо, скільки буде повторів, але відомо, **коли зупинитися**.

````python
cups = 0
while cups < 3:
    cups += 1
    print("Наливаю чашку", cups)
print("Готово!")
````

````python
# Скільки місяців відкладати по 150 грн на кавоварку за 1000 грн?
savings = 0
months = 0
while savings < 1000:
    savings += 150
    months += 1
print(f"Через {months} міс. буде {savings} грн")
````

⚠️ **Нескінченний цикл:** якщо всередині `while` нічого не змінює умову, цикл ніколи не закінчиться. У Colab його зупиняють кнопкою ⏹ біля клітинки (або *Runtime → Interrupt execution*).

## `break` і `continue`

Працюють і в `while`, і в `for`:
- `break` — **негайно вийти** з циклу;
- `continue` — **пропустити** решту коду на цьому кроці й перейти до наступного.

````python
# break: шукаємо перше замовлення дорожче за 200 грн
orders = [120, 85, 240, 65, 310]
for amount in orders:
    if amount > 200:
        print("Знайшли:", amount)
        break                 # далі шукати не треба
````

````python
# continue: рахуємо лише оплачені замовлення (від'ємні — повернення)
payments = [120, -40, 85, -15, 240]
paid = 0
for amount in payments:
    if amount < 0:
        continue              # повернення пропускаємо
    paid += amount
print("Оплачено:", paid)
````

## Кортеж (`tuple`) — незмінний список

Кортеж записують у **круглих** дужках. Він схожий на список, але змінити його не можна. Використовують для даних, що «тримаються разом»: координати, запис про замовлення, пара «назва — ціна».

````python
order = ("A-101", "Лате", 2, 75)
print(order[1])
print(len(order))

# order[1] = "Чай"   # ❌ кортеж змінити не можна: прибери # на початку, запусти й подивись на помилку
````

**Розпакування** — кожен елемент кортежу одразу у свою змінну:

## Множина (`set`) — лише унікальні значення

Множину записують у **фігурних** дужках. У ній немає повторів і немає порядку (тому немає й індексів). Найчастіше її використовують, щоб **прибрати дублікати** або **порівняти дві групи**.

````python
visits = ["Олена", "Андрій", "Олена", "Ірина", "Андрій"]
unique = set(visits)
print(unique, len(unique))

unique.add("Марко")
print("Марко" in unique)
````

Операції з множинами:
- `a | b` — **об'єднання**: усі елементи з обох множин;
- `a & b` — **перетин**: лише спільні елементи;
- `a - b` — **різниця**: є в `a`, але немає в `b`.

````python
monday = {"лате", "еспресо", "чай"}
tuesday = {"лате", "какао", "чай"}

print(monday | tuesday)    # що продавали хоча б в один із днів
print(monday & tuesday)    # що продавали в обидва дні
print(monday - tuesday)    # лише в понеділок
````

⚠️ Порожня множина — це `set()`, а не `{}` (фігурні дужки без нічого — порожній словник, наступна тема).

У клієнта алергія на горіхи й молоко. Визнач:
- `common` — які алергени є в десерті (множина);
- `is_safe` — чи безпечний десерт (`True`, якщо спільних алергенів немає).

<details>

Спільне — перетин `&`. Десерт безпечний, якщо `len(common) == 0`.

````python
allergens = {"горіхи", "молоко"}
dessert = {"борошно", "цукор", "горіхи", "яйця"}

common = allergens & dessert
is_safe = len(common) == 0

print(common, is_safe)
````

## Що таке словник

**Словник** зберігає пари **ключ → значення**, як справжній словник: слово → переклад. Значення шукають не за номером, а **за ключем**.

```
назва = {ключ1: значення1, ключ2: значення2}
```

````python
prices = {"еспресо": 50, "лате": 75, "какао": 60}

print(prices["лате"])          # значення за ключем
print(len(prices))             # кількість пар
print("чай" in prices)         # in перевіряє КЛЮЧІ
````

## Читаємо безпечно: `.get()`

Якщо ключа немає, `prices["чай"]` дасть помилку `KeyError`. Метод `.get(ключ, запасне_значення)` поверне запасне значення замість помилки.

````python
print(prices.get("лате", 0))
print(prices.get("чай", 0))           # ключа немає — отримаємо 0
print(prices.get("чай", "немає в меню"))
````

## Змінюємо словник

| Дія | Код |
|---|---|
| додати нову пару | `prices["капучино"] = 70` |
| змінити значення | `prices["лате"] = 80` |
| збільшити значення | `prices["лате"] += 5` |
| видалити пару | `del prices["какао"]` |

````python
prices = {"еспресо": 50, "лате": 75, "какао": 60}

prices["капучино"] = 70
prices["лате"] += 5
del prices["какао"]

print(prices)
````

## Перебір словника

- `for key in d:` — перебирає **ключі**;
- `d.values()` — лише значення;
- `d.items()` — пари `(ключ, значення)`, які зручно розпакувати.

````python
stock = {"кава": 5, "молоко": 12, "цукор": 3}

for item, amount in stock.items():
    print(f"{item}: {amount} кг")

print("Усього:", sum(stock.values()), "кг")
````

## Список словників — типові «табличні» дані

Кожен словник — один рядок таблиці. Саме так часто виглядають дані з файлів та інтернету.

````python
clients = [
    {"name": "Олена", "orders": 12},
    {"name": "Андрій", "orders": 3},
]

for client in clients:
    print(client["name"], "—", client["orders"], "замовлень")
````

# Тема 11. Фінальний проєкт: звіт кав'ярні за день
