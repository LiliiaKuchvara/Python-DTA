10. Створи три змінні на основі `students`:
- `ref` — просте присвоєння  (b = a);
- `shallow` — поверхнева копія (b = copy(a));
- `deep` — глибока копія (b = copy.deepcopy(a)).

Потім у `shallow` додай студентці `"Sofia"` оцінку 100, а з `ref` видали останнього студента.

Виведи кількість студентів та оцінки студентки `"Sofia"` в `students`, `ref`, `shallow` і `deep`. Поясни в коментарі отримані результати

````python

import copy

# три змінні на основі students
ref = students
shallow = copy.copy(students)
deep = copy.deepcopy(students)

# Додати Софії оцінку 100 у поверхневу копію
shallow[2]["scores"].append(100)

# Видалити останнього студента через ref
ref.pop()

# Виводимо кількість студентів та оцінки Софії
print("students:", len(students), students[2]["scores"])
print("ref:", len(ref), ref[2]["scores"])
print("shallow:", len(shallow), shallow[2]["scores"])
print("deep:", len(deep), deep[2]["scores"])


# ref і students посилаються на один список,
# тому видалення останнього студента через ref
# також видаляє його зі students.
#
# shallow — поверхнева копія зовнішнього списку,
# але словники студентів і вкладені списки оцінок спільні.
# Тому оцінка 100 для Sofia з'явиться у students, ref і shallow.
#
# deep — глибока копія, тому оцінки та дані студентів
# залишаються незалежними від змін в інших змінних.
````

12. Код нижче має:
- вивести кількість відвідувачів у четвер, а якщо даних немає — `0`;
- вивести найбільшу кількість відвідувачів за день;
- перевірити, чи був день, коли прийшло 95 відвідувачів.

Знайди й виправ помилки, поясни кожну в коментарі

````python
visits = {"Mon": 120, "True": 95, "Wed": 140}
print(visits["True"])
print(max(visits))
print(95 in visits)


visits = {"Mon": 120, "True": 95, "Wed": 140}

# 1. Використовуємо get(), щоб отримати кількість відвідувачів у четвер.
# Якщо ключа "True" немає, повертаємо 0 замість помилки KeyError.
print(visits.get("True", 0))

# 2. max(visits.values()) знаходить найбільше значення,
# а max(visits) повернув би назву дня з найбільшим ключем за алфавітом.
print(max(visits.values()))

# 3. Перевіряємо, чи є 95 серед значень словника.
# Оператор in для словника перевіряє ключі, а не значення.
print(95 in visits.values())
````

14. Збережи стан `books` до видачі у змінну `books_before` так, щоб подальші зміни `books` на неї не вплинули.  
Зменш кількість примірників кожної книги на кількість виданих сьогодні.  
Виведи `books_before` і `books`, а також множину книг, які сьогодні не видавалися

````python

import copy

books = {"1984": 3, "Dune": 2, "Atomic Habits": 4, "Sapiens": 1}
borrowed_today = ["Dune", "1984", "Sapiens", "1984"]
readers_mon = {"Iryna", "Maksym", "Oleh"}
readers_tue = {"Maksym", "Sofia"}
# 1.
books_before = copy.deepcopy(books)
# 2.
for book in borrowed_today:
    books[book] -= 1
# 3.
print("До видачі:", books_before)
print("Після видачі:", books)
# 4.
not_borrowed = set(books) - set(borrowed_today)
print("Книги, які не видавалися:", not_borrowed)
````

15. Збери звіт у словник `report` з ключами:
- `"titles"` — скільки різних назв книг є в бібліотеці;
- `"copies_left"` — скільки примірників залишилося загалом;
- `"out_of_stock"` — чи є книга, якої не залишилося в наявності (`True` / `False`);
- `"unique_readers"` — скільки різних читачів приходило за два дні;
- `"loyal_readers"` — множина читачів, які приходили обидва дні.

Виведи `report`

````python

report = {
    "titles": len(books),
    "copies_left": sum(books.values()),
    "out_of_stock": any(amount == 0 for amount in books.values()),
    "unique_readers": len(readers_mon | readers_tue),
    "loyal_readers": readers_mon & readers_tue
}

print(report)
````
