14. Для множин `analyst` і `developer` знайди:
- навички, які є тільки в аналітика
- навички, які є тільки в розробника
- навички, які є лише в одного з них

```python
analyst = {"SQL", "Excel", "Python", "Statistics"}
developer = {"Python", "Git", "SQL", "Docker"}

only_analyst = analyst - developer
only_developer = developer - analyst
only_one = analyst ^ developer

print("Тільки в аналітика:", only_analyst)
print("Тільки в розробника:", only_developer)
print("Лише в одного з них:", only_one)
```

15. Дано два речення:

`s1 = "Python is great for Data Analysis"`  
`s2 = "Data analysis needs python and SQL"`

Виведи, скільки всього унікальних слів у двох реченнях разом, і які слова є тільки в першому реченні.  
Слова, що відрізняються лише регістром (наприклад, `Python` і `python`), вважай однаковими

```python
s1 = "Python is great for Data Analysis"
s2 = "Data analysis needs python and SQL"

words1 = set(s1.lower().split())
words2 = set(s2.lower().split())

total_words = words1.union(words2)
only_in_s1 = words1
print("Кількість унікальних слів:", len(total_words))
print("Слова тільки в першому реченні:", only_in_s1)
```
